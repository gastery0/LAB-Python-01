```python
#1
contador_global = 0

# ERROR: UnboundLocalError porque Python asume que 'contador_global' es una variable local antes de inicializarla.

# FORMA 1
def incrementar_v1():
    global contador_global
    contador_global = contador_global + 1

# FORMA 2
def incrementar_v2(valor):
    return valor + 1

contador_global = incrementar_v2(contador_global)
print(contador_global)

"""
Explicación: La 'FORMA 2' es más fácil de probar (unit testing) porque es una función pura que no depende de estados escondidos fuera de ella. Su salida depende únicamente de su entrada.
"""

#2
def agregar_nota(notas, nueva):
    notas.append(nueva)

def reiniciar(notas):
    # notas = [] reasigna la variable local a un nuevo objeto, no toca la original.
    notas.clear() #SOLUCION CORRECTA usando modificación in-place

mis_notas = [3.5, 4.0]
print("ID original:", id(mis_notas))

agregar_nota(mis_notas, 4.8)
reiniciar(mis_notas)

print("ID final:", id(mis_notas))
print(mis_notas)

#3
cuenta = {"titular": "Ana", "saldo": 200_000, "movimientos": []}

def consultar(cta):
    return cta["saldo"]

def depositar(cta, monto):
    if monto > 0:
        cta["saldo"] += monto
        cta["movimientos"].append(("Depósito", monto, cta["saldo"]))

def retirar(cta, monto):
    if monto > cta["saldo"]:
        print("Error: Fondos insuficientes.")
    elif monto > 0:
        cta["saldo"] -= monto
        cta["movimientos"].append(("Retiro", monto, cta["saldo"]))

while True:
    print("\n--- CAJERO ---")
    print("1. Consultar\n2. Depositar\n3. Retirar\n4. Salir")
    opc = input("Seleccione: ")

    if opc == "1":
        print("Saldo:", consultar(cuenta))
    elif opc == "2":
        m = float(input("Monto a depositar: "))
        depositar(cuenta, m)
    elif opc == "3":
        m = float(input("Monto a retirar: "))
        retirar(cuenta, m)
    elif opc == "4":
        break
    else:
        print("Opción no válida")

print("\n--- EXTRACTO ---")
print(f"{'TIPO':<10} | {'MONTO':<10} | {'SALDO'}")
for mov in cuenta["movimientos"]:
    print(f"{mov[0]:<10} | {mov[1]:<10} | {mov[2]}")
```

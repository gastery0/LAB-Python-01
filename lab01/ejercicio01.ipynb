```python
#1
saldo = float(input("Ingrese el saldo inicial: "))
deposito = float(input("Ingrese el monto a depositar: "))
retiro = float(input("Ingrese el monto a retirar: "))

saldo = saldo + deposito
saldo = saldo - retiro

interes = 0.02
saldo_final = saldo + (saldo * interes)

print(f"Saldo final con interés: ${saldo_final:.2f}")

"""
Respuestas:
- ¿Qué representa el estado inicial y el estado final del programa?
  El estado inicial representa los valores asignados a las variables de memoria antes de ejecutar cualquier operación (el saldo base, el monto a depositar y retirar). El estado final es la instantánea de la memoria después de haber aplicado las operaciones aritméticas y procesado la lógica (el saldo final modificado y calculado con interés).
- Explica cómo cada instrucción afecta la memoria:
  Cada asignación (`saldo = saldo + deposito`) sobrescribe el valor anterior de la variable `saldo` en memoria con el nuevo resultado. Las variables mutan línea tras línea siguiendo la naturaleza del paradigma imperativo.
"""

#2
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]

temperaturas.append(27.0)
temperaturas.pop(0)
temperaturas.insert(1, 23.7)

temperaturas.sort()
print("Lista ordenada:", temperaturas)

diferencia = max(temperaturas) - min(temperaturas)
print("Diferencia (Max - Min):", diferencia)

suma_temp = 0
for t in temperaturas:
    suma_temp += t
promedio = suma_temp / len(temperaturas)
print("Promedio:", promedio)

indice = temperaturas.index(23.7)
print("Índice de 23.7:", indice)

print("Temperaturas de la 2da a la 4ta:", temperaturas[1:4])
print("Tres primeras temperaturas:", temperaturas[:3])

#3
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]

print("Primer producto:", inventario[0]["nombre"])

inventario.append({"nombre": "Huevos", "precio": 500, "stock": 50})

for p in inventario:
    if p["nombre"] == "Leche":
        p["stock"] -= 2

valor_total = 0
for p in inventario:
    valor_total += (p["precio"] * p["stock"])
print("Valor total del inventario:", valor_total)

print("Productos con stock < 10:")
for p in inventario:
    if p["stock"] < 10:
        print(p["nombre"])

for p in inventario:
    p["precio"] = p["precio"] * 1.10
    p["precio"] = round(p["precio"], 2)

buscado = input("Ingrese producto a buscar: ")
encontrado = False
for p in inventario:
    if p["nombre"].lower() == buscado.lower():
        print("Producto encontrado:", p)
        encontrado = True
        break
if not encontrado:
    print("Producto no encontrado en inventario.")

for i in range(len(inventario) - 1, -1, -1):
    if inventario[i]["stock"] == 0:
        inventario.pop(i)

def obtener_precio(prod):
    return prod["precio"]
inventario.sort(key=obtener_precio)
print("Inventario ordenado:", inventario)

#4
import math

rectangulo = (5, 10)
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
print(f"Área: {area}, Perímetro: {perimetro}")

# Intentar cambiar ancho causa error:
# rectangulo[0] = 7
# ¿Por qué? TypeError: 'tuple' object does not support item assignment. Las tuplas son inmutables en memoria.

posiciones = [(0, 0), (1, 2), (3, 4)]
distancia_total = 0

for i in range(len(posiciones) - 1):
    x1 = posiciones[i][0]
    y1 = posiciones[i][1]
    x2 = posiciones[i+1][0]
    y2 = posiciones[i+1][1]
    distancia_paso = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    distancia_total += distancia_paso

print("Distancia total recorrida:", distancia_total)

#5
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

prestados = 0
disponibles = 0
for libro in biblioteca["libros"]:
    if libro["prestado"]:
        prestados += 1
    else:
        disponibles += 1

print(f"Libros prestados: {prestados}, Libros disponibles: {disponibles}")

biblioteca["libros"].append({"titulo": "Cien Años de Soledad", "autor": "Gabriel García Márquez", "prestado": False})

for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True

print("Estado final de la biblioteca:", biblioteca)
```

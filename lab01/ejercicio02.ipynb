```python
#1
edad = int(input("Ingrese su edad: "))
tiene_licencia = input("¿Tiene licencia? (s/n): ").lower()
sancionado = input("¿Está sancionado? (s/n): ").lower()

if edad >= 18 and tiene_licencia == 's' and sancionado == 'n':
    print("Usted puede conducir.")
elif sancionado == 's':
    print("No puede conducir porque está sancionado.")
elif edad < 18:
    print("No puede conducir por ser menor de edad.")
else:
    print("No puede conducir porque le falta la licencia.")

#2
n = int(input("Ingrese un número natural n: "))
suma = 0
for i in range(1, n + 1):
    suma += i
print(f"Suma de los primeros {n} naturales es: {suma}")

print("Números pares:")
for i in range(1, 6):
    if i % 2 == 0:
        print(i)

temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1]
suma_t = 0
for t in temperaturas: suma_t += t
prom_t = suma_t / len(temperaturas)

print("Temperaturas mayores al promedio:")
for t in temperaturas:
    if t > prom_t:
        print(t)

#3
contador = 5
while contador > 0:
    print("Contador:", contador)
    contador -= 1

numero = -1
while numero != 0:
    numero = int(input("Ingrese un número (0 para salir): "))

"""
¿Qué parte representa el estado del ciclo?
La variable 'contador' (en el primer caso) y 'numero' (en el segundo). Sus valores dictan en cada iteración si la condición del 'while' se evalúa como True o False.
"""

#4
estado = {"suma_pares": 0, "suma_impares": 0}

for i in range(1, 11):
    if i % 2 == 0:
        estado["suma_pares"] += i
    else:
        estado["suma_impares"] += i

print("Estado final:", estado)
"""
Explicación del cambio de estado:
En cada iteración, el condicional evalúa si 'i' es divisible por 2. Si lo es, reasigna el valor asociado a la llave 'suma_pares' incrementándolo. Si no, actualiza 'suma_impares'. El diccionario muta paso a paso manteniendo memoria del acumulado.
"""

#5
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre.lower() == "fin":
        break
    precio = float(input("Precio: "))
    cantidad = int(input("Cantidad: "))

    if precio < 0 or cantidad < 0:
        print("Datos inválidos, ignorando producto.")
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]
print("Total del inventario:", total)


if len(productos) > 0:
    def obtener_precio(prod):
        return prod["precio"]
    producto_caro = max(productos, key=obtener_precio)
    print("Producto más caro:", producto_caro)

print("Estado final del inventario:", productos)
```

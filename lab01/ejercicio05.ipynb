```python
#1
import random

ganadas = 0
perdidas = 0

while True:
    objetivo = random.randint(1, 100)
    intentos_restantes = 7
    historial = []
    ganado = False

    print("\nAdivina el número (1-100). Tienes 7 intentos.")

    while intentos_restantes > 0:
        intento = int(input("Tu intento: "))
        historial.append(intento)
        intentos_restantes -= 1

        if intento == objetivo:
            print("¡Ganaste!")
            ganado = True
            ganadas += 1
            break
        elif intento < objetivo:
            print("Mayor...")
        else:
            print("Menor...")

    if not ganado:
        print(f"Perdiste. El número era {objetivo}")
        perdidas += 1

    print("Historial de intentos:", historial)

    jugar_nuevo = input("¿Jugar de nuevo? (s/n): ").lower()
    if jugar_nuevo != 's':
        break

print(f"Marcador: Ganadas {ganadas}, Perdidas {perdidas}")

"""
¿Por qué 7 intentos son suficientes si se juega de forma óptima?
Porque el algoritmo de Búsqueda Binaria parte el espacio a la mitad cada vez. El logaritmo en base 2 de 100 es aprox 6.64. Es decir, con un máximo de 7 particiones a la mitad (2^7 = 128) se puede garantizar encontrar cualquier número entre 1 y 100.
"""

#2
tablero = []
for i in range(10):
    fila = []
    for j in range(10):
        fila.append(0)
    tablero.append(fila)

tablero[5][4] = 1
tablero[5][5] = 1
tablero[5][6] = 1

def contar_vecinos(tab, f, c):
    vecinos = 0
    for i in range(-1, 2):
        for j in range(-1, 2):
            if i == 0 and j == 0:
                continue
            vf = f + i
            vc = c + j
            if 0 <= vf < 10 and 0 <= vc < 10:
                vecinos += tab[vf][vc]
    return vecinos

def imprimir_tablero(tab):
    for i in range(10):
        linea = ""
        for j in range(10):
            if tab[i][j] == 1:
                linea += "■ "
            else:
                linea += "· "
        print(linea)
    print("-" * 20)

for generacion in range(5):
    print(f"Generación {generacion + 1}:")
    imprimir_tablero(tablero)

    nuevo_tablero = []
    for i in range(10):
        fila = []
        for j in range(10):
            fila.append(0)
        nuevo_tablero.append(fila)

    for i in range(10):
        for j in range(10):
            vecinos = contar_vecinos(tablero, i, j)
            if tablero[i][j] == 1:
                if vecinos == 2 or vecinos == 3:
                    nuevo_tablero[i][j] = 1
            else:
                if vecinos == 3:
                    nuevo_tablero[i][j] = 1

    tablero = nuevo_tablero

"""
¿Por qué es necesario usar un tablero nuevo?
Si se modifica el tablero original mientras se recorre, el cambio de estado de las celdas afectaría el conteo de vecinos de las celdas contiguas que aún no han sido procesadas, rompiendo la simulación sincrónica.
"""
```

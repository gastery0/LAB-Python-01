```python
#1
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre.lower() == "fin":
        break

    while True:
        try:
            precio = float(input("Precio: "))
            cantidad = int(input("Cantidad: "))
            break
        except ValueError:
            print("Error: Ingresaste letras o caracteres inválidos. Vuelve a intentar.")
        else:
            print("Este else no se ejecuta por el break, pero normalmente iría aquí si el Try funciona perfectamente.")
        finally:
            print("--> Intentando captura de datos (finally se ejecuta siempre)")

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

#2
import json

inventario_data = [
    {"nombre": "Pan", "precio": 1500.0, "cantidad": 3},
    {"nombre": "Leche", "precio": 3000.0, "cantidad": 2}
]

archivo = open("inventario.csv", "w")
for p in inventario_data:
    archivo.write(f"{p['nombre']},{p['precio']},{p['cantidad']}\n")
archivo.close()

try:
    archivo_leer = open("inventario.csv", "r")
    inventario_csv = []
    for linea in archivo_leer:
        partes = linea.strip().split(",")
        inventario_csv.append({
            "nombre": partes[0],
            "precio": float(partes[1]),
            "cantidad": int(partes[2])
        })
    archivo_leer.close()
    print("CSV cargado:", inventario_csv)
except FileNotFoundError:
    inventario_csv = []
    print("Archivo CSV no existe, inventario vacío.")

with open("inventario.json", "w") as j_file:
    json.dump(inventario_data, j_file)

try:
    with open("inventario.json", "r") as j_file:
        inventario_json = json.load(j_file)
        print("JSON cargado:", inventario_json)
except FileNotFoundError:
    inventario_json = []
    print("Archivo JSON no existe.")

"""
¿Qué ventajas tiene JSON frente al CSV?
- JSON respeta directamente la estructura de diccionarios de Python y conserva los tipos de datos (como float e integer) sin requerir conversiones manuales mediante 'split' y 'float()'. CSV solo maneja texto plano.
"""
```

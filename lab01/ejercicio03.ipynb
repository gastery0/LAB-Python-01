```python
#1
import requests

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()

nombres_biz = []
for u in usuarios:
    if u["email"].endswith(".biz"):
        nombres_biz.append(u["name"])

print("Usuarios con correo .biz encontrados:")
for n in nombres_biz:
    print("-", n)

#2
nombre_pokemon = input("Ingrese el nombre de un Pokémon: ").lower()
url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
respuesta = requests.get(url)

if respuesta.status_code == 200:
    datos = respuesta.json()

    tipos_lista = []
    for t in datos["types"]:
        tipos_lista.append(t["type"]["name"])

    habilidades_lista = []
    for h in datos["abilities"]:
        habilidades_lista.append(h["ability"]["name"])

    pokemon = {
        "nombre": datos["name"],
        "altura": datos["height"],
        "peso": datos["weight"],
        "tipos": tipos_lista,
        "habilidades": habilidades_lista,
        "experiencia_base": datos["base_experience"]
    }

    print("\n📘 Información del Pokémon")
    print(f"Nombre: {pokemon['nombre'].capitalize()}")
    print(f"Altura: {pokemon['altura']}")
    print(f"Peso: {pokemon['peso']}")

    if len(pokemon["tipos"]) > 1:
        print("Tipos: Pokémon de tipo mixto")
    else:
        print(f"Tipos: {pokemon['tipos'][0]}")

    if pokemon["peso"] > 100:
        print("Clasificación: Pokémon de gran tamaño")
    else:
        print("Clasificación: Pokémon ligero")

    print("Habilidades:")
    for hab in pokemon["habilidades"]:
        print(f" - {hab}")

    print(f"Experiencia base obtenida al derrotarlo: {pokemon['experiencia_base']}")

else:
    print("Pokémon no encontrado. Verifique el nombre ingresado.")
```
